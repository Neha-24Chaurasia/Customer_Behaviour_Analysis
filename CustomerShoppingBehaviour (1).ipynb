```python
# Lets fill the missing values using median . But we will take median category wise not overall .
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
    
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})

    
```

```python
# Creating age_group column 
labels= ['Young Adult', 'Adult', 'Middle_aged', 'Senior']
df['age_group']=pd.qcut(df['age'], q=4, labels=labels)
```

```python
df['frequency_of_purchases'].value_counts()
```

```python
# Doing mapping of frequency_of_purchases
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly': 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days']= df['frequency_of_purchases'].map(frequency_mapping )
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
# As 2 pair of columns are same so we can replace it so instead of 7 labels we have 5 labels now
df['frequency_of_purchases'] = df['frequency_of_purchases'].replace({
    'Fortnightly': 'Bi-Weekly',
    'Every 3 Months': 'Quarterly'
})
```

```python
# As both the columns are same lets drop oone column
df=df.drop('promo_code_used', axis=1)
```

```python
df.columns
```

```python
num_cols = ['age', 'purchase_amount', 'previous_purchases', 'review_rating']

for col in num_cols:
    q1, q3 = df[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    outliers = df[(df[col] < q1 - 1.5*iqr) | (df[col] > q3 + 1.5*iqr)]
    print(col, "->", len(outliers), "outliers")

df[num_cols].boxplot()
```

#### “Outliers were checked using the IQR method and box plots. No outliers were found in age, purchase amount, previous purchases or review rating.”

```python
from sklearn.preprocessing import RobustScaler
from sklearn.cluster import KMeans

features = ['previous_purchases', 'purchase_amount', 'review_rating']
X = RobustScaler().fit_transform(df[features])

df['customer_group'] = KMeans(n_clusters=4, random_state=42, n_init=10).fit_predict(X)

df.groupby('customer_group')[features].mean().round(2)
```
